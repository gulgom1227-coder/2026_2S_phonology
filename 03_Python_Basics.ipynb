### For
```python
# Syntax
for item in sequence:
	# execute this code block for each item
```

````python
a = [1,2,3,4]

print(a[0])
print(a[1])
print(a[2])
print(a[3])

for i in a: #i는 임시 변수, a는 출력을 원하는 데이터(출처)
  print(i)
````

````python
# 예시: 플레이리스트 재생하기

# 플레이리스트 안에 있는 곡을 차례로 하나씩 재생한다.

playlist = ['gedo', 'hedgehog', 'culture', 'BYB']

for play in playlist:
  print(play)  # play는 BYB로 저장됨
````

````python
#  예시: 일주일 걸음 수 목표 확인하기
daily_steps = [7200, 10500, 8300, 12400, 10000, 6500, 11300]

goal = 10000
success_days = 0

for days in daily_steps:
  if days >= goal:
    success_days = success_days + 1
  else:
    success_days = success_days + 0

print(success_days)


# 일주일 동안 목표를 달성한 일수를 계산한다.
````

### While
```python
# Syntax
while condition:
	# execute this code block as long as the condition is true
```

````python
# 예시: 카페 쿠폰 스탬프 채우기
stamp = 0
target = 10

while stamp < target:
  print(stamp)
  stamp = stamp + 1

print(stamp)

# 스탬프가 10개가 될 때까지 스탬프를 적립한다.
````

````python
a = 1
while a < 5:
  a = a + 1
  print(a)
````
